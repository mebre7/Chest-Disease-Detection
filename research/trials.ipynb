## `ConfigBox`
- `ConfigBox` is a class that provides a convenient way to access configuration parameters from a dictionary-like object. It allows you to retrieve values using dot notation, making it easier to work with nested configurations.

```python
import yaml
file_path = 'test.yaml'

with open(file_path, 'r') as file:
    data = yaml.safe_load(file)

print(data)
```

```python
from box import ConfigBox

config_data = ConfigBox(data)
print(config_data.name)
```

## `@ensure_annotations`
- `@ensure_annotations` is a decorator that can be used to enforce type annotations on function parameters and return values. It helps ensure that the function is called with the correct types, improving code readability and reducing potential bugs. It is from the `ensure` library, which provides runtime type checking for Python functions.

```python
from ensure import ensure_annotations

@ensure_annotations
def multiply_numbers(a: int, b: int) -> int:
    return a * b
```

```python
multiply_numbers(5, 10)
```

```python
multiply_numbers(5, '10')
```
